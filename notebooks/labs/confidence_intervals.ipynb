Calculate a simple t interval for fictional observations.

```python
import numpy as np
from scipy import stats
x=np.array([5.1,4.9,5.4,5.0,5.2,4.8,5.3,5.1])
mean=x.mean(); se=stats.sem(x)
(mean, stats.t.interval(.95,len(x)-1,loc=mean,scale=se))
```
