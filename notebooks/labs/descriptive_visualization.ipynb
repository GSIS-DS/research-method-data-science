```python
import pandas as pd
import matplotlib.pyplot as plt
url='https://raw.githubusercontent.com/GSIS-DS/research-method-data-science/main/data/sample/mini_country_indicators.csv'
df=pd.read_csv(url)
df.select_dtypes(include='number').describe()
```

Plot the six observations; describe without causal language.

```python
ax=df.plot.scatter(x='income_per_person',y='life_expectancy')
ax.set(title='Fictional teaching data',xlabel='Income per person (arbitrary units)',ylabel='Life expectancy (years)')
plt.show()
```
