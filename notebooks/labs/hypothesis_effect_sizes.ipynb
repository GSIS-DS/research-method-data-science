Compare two fictional groups and calculate a standardized difference.

```python
import numpy as np
from scipy import stats
a=np.array([4,5,6,5,7,6]); b=np.array([3,4,4,5,3,4])
test=stats.ttest_ind(a,b,equal_var=False)
pooled=np.sqrt((a.var(ddof=1)+b.var(ddof=1))/2)
d=(a.mean()-b.mean())/pooled
(test,d)
```
