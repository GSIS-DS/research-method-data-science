Build two small transparent tables.

```python
import pandas as pd
left=pd.DataFrame({'id':[1,2,3],'score':[8,None,6]})
right=pd.DataFrame({'id':[1,2,3],'group':['A','A','B']})
print(left.isna().sum())
```

Join and check whether row count and keys behave as expected.

```python
joined=left.merge(right,on='id',validate='one_to_one')
print(joined)
assert len(joined)==len(left)
```
