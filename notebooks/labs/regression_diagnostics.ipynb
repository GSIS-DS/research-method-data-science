```python
import pandas as pd
import statsmodels.formula.api as smf
url='https://raw.githubusercontent.com/GSIS-DS/research-method-data-science/main/data/sample/mini_country_indicators.csv'
df=pd.read_csv(url)
m=smf.ols('life_expectancy ~ income_per_person',df).fit()
m.get_influence().summary_frame()[['standard_resid','cooks_d']]
```
