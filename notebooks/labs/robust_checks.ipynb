Compare a baseline and region-adjusted descriptive model.

```python
import pandas as pd
import statsmodels.formula.api as smf
url='https://raw.githubusercontent.com/GSIS-DS/research-method-data-science/main/data/sample/mini_country_indicators.csv'
df=pd.read_csv(url)
base=smf.ols('life_expectancy ~ income_per_person',df).fit()
adj=smf.ols('life_expectancy ~ income_per_person + C(region)',df).fit()
{'baseline':base.params.to_dict(),'adjusted':adj.params.to_dict()}
```
