Evaluate a small predictive split while noting that accuracy does not establish causality.

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error
X=[[1],[2],[3],[4],[5],[6]]; y=[2,4,5,8,9,12]
m=DecisionTreeRegressor(max_depth=2,random_state=3002).fit(X[:4],y[:4])
mean_absolute_error(y[4:],m.predict(X[4:]))
```
